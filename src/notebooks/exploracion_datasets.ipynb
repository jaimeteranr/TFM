```python
df_horarios = pd.read_excel("horarios.xlsx")

# Asegurar formato datetime
df_horarios["entrada"] = pd.to_datetime(
    df_horarios["entrada"].astype(str),
    format="%H:%M:%S"
)

# Redondear a medias horas
df_horarios["entrada_norm"] = (
    df_horarios["entrada"]
    .dt.round("30min")
    .dt.strftime("%H:%M")
)

# Redondear duración a múltiplos de 0.5h
df_horarios["duracion_norm"] = (
    df_horarios["duracion_turno"] * 2
).round() / 2

turnos = (
    df_horarios
    .groupby(["entrada_norm", "duracion_norm"])
    .size()
    .reset_index(name="frecuencia")
    .sort_values("frecuencia", ascending=False)
)

print(turnos.head(50))

total_turnos = turnos["frecuencia"].sum()
print(f"Total turnos: {total_turnos}")
```

```python
print(turnos.shape[0])
```

```python
top10 = turnos.head(10)["frecuencia"].sum()
top20 = turnos.head(20)["frecuencia"].sum()
total = turnos["frecuencia"].sum()

print("Top 10:", round(100*top10/total,2), "%")
print("Top 20:", round(100*top20/total,2), "%")
```

```python
print(turnos.head(30))

turnos["acumulado"] = (
    turnos["frecuencia"]
    .cumsum()
    /
    turnos["frecuencia"].sum()
)

print(
    turnos[
        ["entrada_norm",
         "duracion_norm",
         "frecuencia",
         "acumulado"]
    ].head(30)
)
```

```python
# Fecha
df_horarios["fecha"] = pd.to_datetime(
    df_horarios["fecha"]
)

# Hora de entrada
df_horarios["entrada"] = pd.to_datetime(
    df_horarios["entrada"],
    errors="coerce"
)

# ----------------------------------------
# EXPANDIR TURNOS A BLOQUES DE 30 MINUTOS
# ----------------------------------------

registros = []

for _, row in df_horarios.iterrows():

    fecha = row["fecha"]

    inicio = pd.to_datetime(
    row["entrada_norm"],
    format="%H:%M"
)
    fin = inicio + pd.to_timedelta(
        row["duracion_norm"],
        unit="h"
    )

    instante = inicio

    while instante < fin:

        registros.append(
            {
                "fecha": fecha,
                "hora": instante.strftime("%H:%M")
            }
        )

        instante += timedelta(minutes=30)

cobertura = pd.DataFrame(registros)

# ----------------------------------------
# PERSONAS POR FECHA Y MEDIA HORA
# ----------------------------------------

cobertura_real = (
    cobertura
    .groupby(
        ["fecha", "hora"]
    )
    .size()
    .reset_index(name="personas")
)

print("\nCOBERTURA REAL")
print(cobertura_real.head(50))

# ----------------------------------------
# DIA SEMANA
# ----------------------------------------

cobertura_real["dia_semana"] = (
    cobertura_real["fecha"]
    .dt.day_name()
)

cobertura_media = (
    cobertura_real
    .groupby(["dia_semana", "hora"])["personas"]
    .mean()
    .reset_index()
)

# Orden correcto de días
orden = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

cobertura_media["dia_semana"] = pd.Categorical(
    cobertura_media["dia_semana"],
    categories=orden,
    ordered=True
)

cobertura_media = cobertura_media.sort_values(
    ["dia_semana", "hora"]
)

print("\nCOBERTURA MEDIA")
print(cobertura_media.head(200))

# =========================
# TABLA PIVOT MUY ÚTIL
# =========================

pivot = cobertura_media.pivot(
    index="hora",
    columns="dia_semana",
    values="personas"
)

print("\nPIVOT COBERTURA")
print(pivot.round(2))
```
